# Q1 Prediction of fraudulent transactions

1) Write down the probability space:
$\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$
$\mathfrak{F} = \mathcal{P}(\Omega)$
$P(\{\omega_1\}) = 0.50, P(\{\omega_2\}) = 0.10, P(\{\omega_3\}) = 0.20, P(\{\omega_4\}) = 0.20$

2) Compute the $\sigma$-algebra generated by $Y$:
$\sigma(Y) = \{\emptyset, \Omega, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}\}$

3) Explain why $E[X \mid Y]$ must be $\sigma(Y)$-measurable:
Conditional expectation projects a random variable onto the available information. Being measurable with respect to $\sigma(Y)$ means it assigns a constant value to every outcome within the partitions generated by $Y$.

4) Compute $E[X \mid Y]$ explicitly:
$E[X \mid Y=0] = \frac{0(0.50) + 1(0.10)}{0.60} = \frac{1}{6}$
$E[X \mid Y=1] = \frac{0(0.20) + 1(0.20)}{0.40} = \frac{1}{2}$
$E[X \mid Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}$

5) Verify the defining property for every set $A \in \sigma(Y)$:
For $A = \{\omega_1, \omega_2\}$: $\int_A E[X \mid Y] dP = \frac{1}{6}(0.60) = 0.10$. $\int_A X dP = 0(0.50) + 1(0.10) = 0.10$.
For $A = \{\omega_3, \omega_4\}$: $\int_A E[X \mid Y] dP = \frac{1}{2}(0.40) = 0.20$. $\int_A X dP = 0(0.20) + 1(0.20) = 0.20$.
For $A = \Omega$: $0.10 + 0.20 = 0.30$. $\int_\Omega X dP = 0.30$.
For $A = \emptyset$: $0 = 0$.

6) Find the Conditional PMF $p_{X \mid Y}(x \mid y)$:
$y=0: p_{X \mid Y}(0 \mid 0) = \frac{5}{6}, p_{X \mid Y}(1 \mid 0) = \frac{1}{6}$
$y=1: p_{X \mid Y}(0 \mid 1) = \frac{1}{2}, p_{X \mid Y}(1 \mid 1) = \frac{1}{2}$

7) Verify the identity:
$y=0: \mathbb{E}[X \mid Y=0] = 0(\frac{5}{6}) + 1(\frac{1}{6}) = \frac{1}{6}$
$y=1: \mathbb{E}[X \mid Y=1] = 0(\frac{1}{2}) + 1(\frac{1}{2}) = \frac{1}{2}$

8) Calculate the Mean Square Error:
$MSE = \sum (x_i - \hat{x})^2 p_i$
$\omega_1: (0 - \frac{1}{6})^2 \times 0.50 = \frac{1}{72}$
$\omega_2: (1 - \frac{1}{6})^2 \times 0.10 = \frac{5}{72}$
$\omega_3: (0 - \frac{1}{2})^2 \times 0.20 = \frac{1}{20}$
$\omega_4: (1 - \frac{1}{2})^2 \times 0.20 = \frac{1}{20}$
$MSE = \frac{6}{72} + \frac{2}{20} = \frac{11}{60} \approx 0.1833$

9) Mean Square Error if ignored $Y$:
$MSE = Var(X) = E[X^2] - (E[X])^2 = 0.30 - 0.09 = 0.21$
Using the global average yields the full variance of $0.21$. Utilizing $Y$ drops the error to $\frac{11}{60}$. The difference is the variance of the conditional expectation itself according to the Law of Total Variance.

10) Entropy Calculations:

b) $H(X \mid Y=0) = - \frac{5}{6} \log_2(\frac{5}{6}) - \frac{1}{6} \log_2(\frac{1}{6}) \approx 0.6500$
$H(X \mid Y=1) = - \frac{1}{2} \log_2(\frac{1}{2}) - \frac{1}{2} \log_2(\frac{1}{2}) = 1$
$H(X \mid Y) = 0.60(0.6500) + 0.40(1) = 0.7900$
c) $I(X; Y) = H(X) - H(X \mid Y) = 0.8813 - 0.7900 = 0.0913$
d) $\frac{0.0913}{0.8813} \approx 10.36\%$ removed

# Q2 Conditional Density of the Multivariate Gaussian

1) Show the conditional distribution:
Define $X' = X - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}Y$.
$Cov(X', Y) = Cov(X, Y) - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}Cov(Y, Y) = \boldsymbol{\Sigma}_{XY} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YY} = 0$.
Since $X'$ and $Y$ are jointly normal and uncorrelated, they are independent.
$E[X'] = \boldsymbol{\mu}_X - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\mu}_Y$.
$Var(X') = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} = \boldsymbol{\Sigma}_{X \mid Y}$.
Substitute $X = X' + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}Y$ given $Y=y$:
$E[X \mid Y=y] = E[X'] + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}y = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(y - \boldsymbol{\mu}_Y) = \boldsymbol{\mu}^*$.
$Var(X \mid Y=y) = Var(X') = \boldsymbol{\Sigma}_{X \mid Y}$.
Since $X \mid Y=y$ is a linear transformation of a normal variable, it is multivariate normal with mean $\boldsymbol{\mu}^*$ and covariance $\boldsymbol{\Sigma}_{X \mid Y}$.

# Q3 Continuous Fault Monitoring of a Spindle Motor

1) Joint Distribution:
$E[Y] = E[X+Q] = 0$.
$Var(Y) = Var(X) + Var(Q) = 1 + 4 = 5$.
Since $Y$ is a linear combination of independent normal variables, $Y \sim \mathscr{N}(0, 5)$.
$Cov(X, Y) = Cov(X, X+Q) = Cov(X, X) + Cov(X, Q) = Var(X) + 0 = 1$.

2) Information Geometry:
$X$ is not $\sigma(Y)$-measurable because the mapping is not invertible. A reading of $Y=5$ corresponds to infinite $X+Q=5$ combinations. The exact value of $X$ cannot be uniquely determined from $Y$ alone.

3) Conditional Expectation:
$E[X \mid Y] = 0 + \frac{1}{5}(Y - 0) = \frac{1}{5}Y$.
$E[X \mid Y] = \frac{Var(X)}{Var(X) + Var(Q)}Y$.

4) Verification of the Residual:
$Z = X - \frac{1}{5}Y$.
$E[Z] = E[X] - \frac{1}{5}E[Y] = 0 - 0 = 0$.
$Cov(Z, Y) = Cov(X - \frac{1}{5}Y, Y) = Cov(X, Y) - \frac{1}{5}Var(Y) = 1 - \frac{1}{5}(5) = 0$.

5) Information Theory Differential Entropy:
$h(X) = \frac{1}{2} \ln(2\pi e Var(X)) = \frac{1}{2} \ln(2\pi e) \approx 1.4189$
$\sigma_{X \mid Y}^2 = 1 - \frac{1^2}{5} = 0.8$.
$h(X \mid Y) = \frac{1}{2} \ln(2\pi e (0.8)) \approx 1.2958$

6) Mutual Information:
$I(X; Y) = h(X) - h(X \mid Y) = \frac{1}{2} \ln(2\pi e) - \frac{1}{2} \ln(2\pi e (0.8)) = \frac{1}{2} \ln(1.25) \approx 0.1115$
Percentage removed: $\frac{0.1115}{1.4189} \approx 7.86\%$.
If $Var(Q) = 100$: $Var(Y) = 101$. $\sigma_{X \mid Y}^2 = 1 - \frac{1}{101} = \frac{100}{101}$. $I(X; Y) = \frac{1}{2} \ln(1.01) \approx 0.0049$.

7) Diagnostic Interpretation:
$Y$ is a biased estimator because the variance of $Q$ is much larger than the variance of $X$. A high reading of $Y$ is more likely caused by the noise $Q$ than the signal $X$. The conditional expectation $E[X \mid Y] = \frac{1}{5}Y$ shrinks the estimate toward the zero mean to adjust for this noise.
