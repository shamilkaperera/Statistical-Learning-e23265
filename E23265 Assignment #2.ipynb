# Q3) Transformer Failures

1) Construct a probability space:
- $\Omega = \{0, 1, 2, 3\}$.
- The event space $\mathfrak{F}$ is the power set of $\Omega$, meaning all possible subsets of $\{0, 1, 2, 3\}$.
- $P(\{0\})=0.50$, $P(\{1\})=0.30$, $P(\{2\})=0.15$, $P(\{3\})=0.05$.

2) Define $X$ as a random variable:
- The random variable $X: \Omega \to \mathbb{R}$ is defined as the identity function $X(\omega) = \omega$ for all $\omega \in \Omega$.

3) Probability mass function (PMF):
- $p_X(0) = 0.50$
- $p_X(1) = 0.30$
- $p_X(2) = 0.15$
- $p_X(3) = 0.05$
- $p_X(x) = 0$ otherwise.

4) Compute the expectation $E[X]$:
$$E[X] = \sum x \cdot P(X=x) = (0)(0.50) + (1)(0.30) + (2)(0.15) + (3)(0.05) = 0.75$$

5) Compute the variance $\mathrm{Var}(X)$:
$$E[X^2] = (0^2)(0.50) + (1^2)(0.30) + (2^2)(0.15) + (3^2)(0.05) = 1.35$$
$$\mathrm{Var}(X) = E[X^2] - (E[X])^2 = 1.35 - (0.75)^2 = 0.7875$$

6) Find probabilities:
$$P(X \ge 1) = 1 - P(X=0) = 1 - 0.50 = 0.50$$
$$P(X \ge 2) = P(X=2) + P(X=3) = 0.15 + 0.05 = 0.20$$

7) Practical meaning:
- The expectation (0.75) represents the average number of transformer failures expected per day over a long period. The variance (0.7875) quantifies the uncertainty or daily fluctuation around this average expectation.

# Q4) Probability of Defective Items

1) Construct a probability space:
- $\Omega = [0, 1]$, representing all possible proportions of defective items.
- $\mathfrak{F} = \mathcal{B}([0, 1])$, the Borel $\sigma$-algebra on $[0,1]$.
- $P(A) = \int_A f(x)dx$, using the Beta density function for probabilities.

2) Define $X$ as a continuous random variable:
- $X: \Omega \to \mathbb{R}$ such that $X(\omega) = \omega$.

3) Probability density function of $X$:
$$f(x) = \frac{1}{B(2,18)}x^1(1-x)^{17} \quad \text{for } 0 < x < 1, \text{ and } 0 \text{ otherwise}$$

4) Verify $f$ is a valid probability density function:
The integral of the density over its support is $\int_0^1 \frac{1}{B(2,18)} x^1(1-x)^{17} dx$. Since $B(2,18) = \int_0^1 t^{2-1}(1-t)^{18-1} dt$, the integral evaluates to $\frac{B(2,18)}{B(2,18)} = 1$.

5) Compute expectation $E[X]$:
For a Beta distribution, $E[X] = \frac{\alpha}{\alpha+\beta}$.
$$E[X] = \frac{2}{2+18} = \frac{2}{20} = 0.10$$

6) Compute variance $\mathrm{Var}(X)$:
For a Beta distribution, $\mathrm{Var}(X) = \frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$.
$$\mathrm{Var}(X) = \frac{2 \times 18}{(20)^2(21)} = \frac{36}{8400} = \frac{3}{700} \approx 0.004286$$

7) Probability defective $\le 0.10$:
$$P(X \le 0.10) = \int_0^{0.10} \frac{1}{B(2,18)} x(1-x)^{17} dx$$

8) Probability defective between $0.05$ and $0.15$:
$$P(0.05 \le X \le 0.15) = \int_{0.05}^{0.15} \frac{1}{B(2,18)} x(1-x)^{17} dx$$

9) Why a Beta distribution is a reasonable model:
- The Beta distribution is bounded between 0 and 1, mapping perfectly to proportions. Its flexible shape allows it to model processes where defects are typically low.

10) Interpret expectation and variance:
- Expectation (0.10) dictates that, on average, 10% of items are defective in a batch. The variance measures the day-to-day consistency of the manufacturing defect rate.

# Q5) Expected Project Completion Time

1) Specify the support of $X$:
- The support is $[a, b] = [4, 10]$.

2) Write down the PDF $f_X(x)$:
$$f_X(x) = \begin{cases} \frac{x-4}{6} & 4 \le x \le 6 \\ \frac{10-x}{12} & 6 < x \le 10 \\ 0 & \text{otherwise} \end{cases}$$

3) Verify that $f_X$ integrates to 1:
The area of the triangle is $\frac{1}{2} \times \text{base} \times \text{height}$. The base is $10-4=6$. The height at the peak $c=6$ is $\frac{2}{10-4} = \frac{1}{3}$. Area = $\frac{1}{2} \times 6 \times \frac{1}{3} = 1$.

4) Compute $E[X]$:
$$E[X] = \frac{a+b+c}{3} = \frac{4+10+6}{3} = \frac{20}{3} \approx 6.67 \text{ hours}$$

5) Compute $\mathrm{Var}(X)$:
$$\mathrm{Var}(X) = \frac{a^2+b^2+c^2-ab-ac-bc}{18} = \frac{16+100+36-40-24-60}{18} = \frac{14}{9} \approx 1.556$$

6) Interpret $a, b, c$:
- a=4: Optimistic (minimum) completion time.
- b=10: Pessimistic (maximum) completion time.
- c=6: Most likely (mode) completion time.

7) Why Triangular is reasonable:
- It intuitively translates expert estimates into a mathematically closed probability distribution without requiring exact historical data.

8) Compute $P(X \le 5)$:
$$P(X \le 5) = \int_4^5 \frac{x-4}{6} dx = \left[\frac{(x-4)^2}{12}\right]_4^5 = \frac{1}{12} \approx 0.0833$$

9) Compute $P(5 \le X \le 8)$:
$$P(X > 8) = \int_8^{10} \frac{10-x}{12} dx = \left[-\frac{(10-x)^2}{24}\right]_8^{10} = \frac{4}{24} = \frac{1}{6}$$
$$P(5 \le X \le 8) = 1 - P(X < 5) - P(X > 8) = 1 - \frac{1}{12} - \frac{2}{12} = \frac{9}{12} = 0.75$$

10) Practical use:
- Can be used to generate simulated completion scenarios to quantify schedule risk and optimize resource planning.

# Q6) Expected Project Completion Time Using PERT

1) Compute $\alpha$ and $\beta$:
Given $a=8$, $b=20$, $c=11$.
$$\alpha = 1 + 4\left(\frac{11-8}{20-8}\right) = 1 + 4\left(\frac{3}{12}\right) = 2$$
$$\beta = 1 + 4\left(\frac{20-11}{20-8}\right) = 1 + 4\left(\frac{9}{12}\right) = 4$$

2) Support of $X$:
- $[a, b] = [8, 20]$.

3) PDF $f_X(x)$:
$$f_X(x) = \frac{(x-8)^{1}(20-x)^{3}}{B(2,4)(12)^{5}} \quad \text{for } 8 \le x \le 20$$

4) State probability space:
- $\Omega = [8, 20]$, $\mathfrak{F} = \mathcal{B}([8,20])$, $P(A) = \int_A f(x)dx$.

5) Define $X$:
- $X: \Omega \to \mathbb{R}$ with $X(\omega) = \omega$.

6) Show expectation formula:
The mean of the internal Beta component is $\frac{\alpha}{\alpha+\beta}$. Given $\alpha + \beta = 2 + 4\frac{(c-a)+(b-c)}{b-a} = 6$.
$$E[X] = a + (b-a)\left(\frac{\alpha}{6}\right) = a + (b-a)\frac{1 + 4\frac{c-a}{b-a}}{6} = \frac{6a + b - a + 4c - 4a}{6} = \frac{a + 4c + b}{6}$$

7) Compute variance using approximation:
$$E[X] = \frac{8 + 4(11) + 20}{6} = \frac{72}{6} = 12$$
$$\mathrm{Var}(X) = \frac{(12-8)(20-12)}{7} = \frac{4 \times 8}{7} = \frac{32}{7} \approx 4.57$$

8) Compute standard deviation $\sigma$:
$$\sigma = \sqrt{\frac{32}{7}} \approx 2.138$$

9) PERT vs Triangular:
- A PERT distribution concentrates more probability mass around the peak resulting in a curve that looks more like a bell curve. This makes extreme delays less probable, more closely mimicking actual project outcomes.

10) Interpret values:
- a, b, c dictate the optimistic, pessimistic, and expected bounds of a task. $E[X]$ gives the weighted expected time for planning, while $\sigma$ quantifies the standard schedule risk.

# Q7) If the PERT Expectation is to Match Our Intuition

$$11 = \frac{5 + 4(11) + b}{6}$$
