1.

The conditional density is defined by

$$
f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y})=
\frac{f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}
$$

The joint Gaussian density is

$$
f_{\mathbf{X},\mathbf{Y}}(\mathbf{z})=
\frac{1}{\sqrt{(2\pi)^{n+m}|\Sigma|}}
\exp\left[-\frac12(\mathbf{z}-\mu)^T\Sigma^{-1}(\mathbf{z}-\mu)\right]
$$

Partition the mean and covariance matrix as

$$
\mu=
\begin{bmatrix}
\mu_X \
\mu_Y
\end{bmatrix}
$$

and

$$
\Sigma=
\begin{bmatrix}
\Sigma_{XX} & \Sigma_{XY} \
\Sigma_{YX} & \Sigma_{YY}
\end{bmatrix}
$$

Then the conditional distribution of $\mathbf{X}|\mathbf{Y}=\mathbf{y}$ is Gaussian with mean

$$
\mu_{X|Y}=\mu_X+\Sigma_{XY}\Sigma_{YY}^{-1}(\mathbf{y}-\mu_Y)
$$

and covariance

$$
\Sigma_{X|Y}=\Sigma_{XX}-\Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}
$$

Hence

$$
f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y})=
\frac{1}{\sqrt{(2\pi)^n|\Sigma_{X|Y}|}}
\exp\left[-\frac12(\mathbf{x}-\mu_{X|Y})^T
\Sigma_{X|Y}^{-1}
(\mathbf{x}-\mu_{X|Y})
\right]
$$

Therefore the conditional distribution is multivariate normal.

Cov(X,Y)=Cov(X,X+Q)

=Var(X)+0=1
$$

$$
(X,Y)
$$

is jointly Gaussian with mean vector

$$
\mu=
\begin{bmatrix}
0 \
0
