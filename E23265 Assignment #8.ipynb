The Partition FunctionThe canonical partition function is given by:$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{E_j}{k_B T}}$$Substituting the energy eigenvalues $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$ and defining $\beta = \frac{1}{k_B T}$:$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega \left(j + \frac{1}{2}\right)} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^{\infty} \left(e^{-\beta \hbar \omega}\right)^j$$Since $e^{-\beta \hbar \omega} < 1$, the summation is a convergent infinite geometric series taking the form $\sum_{j=0}^{\infty} r^j = \frac{1}{1 - r}$.$$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \left( \frac{1}{1 - e^{-\beta \hbar \omega}} \right) = \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$2. Thermodynamic PotentialsHelmholtz Free Energy, $F(T) = -k_B T \ln Z$:$$F(T) = -k_B T \left( -\frac{\hbar \omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right)$$$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$Internal Energy, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$:$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$$$U(T) = -\frac{\partial}{\partial \beta} \left( -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right) = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$3. EntropyUsing $S = -\left(\frac{\partial F}{\partial T}\right)$ and taking the derivative of $F(T)$:$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \left( \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \right) \left( -e^{-\frac{\hbar \omega}{k_B T}} \right) \left( \frac{\hbar \omega}{k_B T^2} \right)$$$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \left( \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right)$$Verification of $U = F + TS$:$$F + TS = \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)} \right]$$$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U(T)$$4. Heat Capacity and Hyperbolic FormHeat Capacity, $C_V$:$$C_V = \frac{\partial U}{\partial T} = \frac{\partial}{\partial T} \left( \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right) = -\frac{\hbar \omega}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} \left( e^{\frac{\hbar \omega}{k_B T}} \right) \left( -\frac{\hbar \omega}{k_B T^2} \right)$$$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$Hyperbolic Form: Multiply numerator and denominator by $e^{-\frac{\hbar \omega}{k_B T}}$:$$\frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} = \frac{1}{e^{-\frac{\hbar \omega}{k_B T}}\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} = \frac{1}{\left(e^{\frac{\hbar \omega}{2k_B T}} - e^{-\frac{\hbar \omega}{2k_B T}}\right)^2}$$Since $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we know $(e^x - e^{-x})^2 = 4 \sinh^2(x)$. Thus:$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{4 \sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)} = k_B \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$5. The Maximum Entropy Distribution under Mean and Variance ConstraintsWe want to find $p(x)$ that maximizes the differential entropy $H = -\int p(x) \ln p(x) dx$ subject to three constraints:$\int p(x) dx = 1$$\int x p(x) dx = \mu$$\int (x-\mu)^2 p(x) dx = \sigma^2$We set up the Lagrangian with multipliers $\lambda_0, \lambda_1, \lambda_2$:$$\mathcal{L} = -\int p(x) \ln p(x) dx - (\lambda_0 - 1)\left(\int p(x) dx - 1\right) - \lambda_1\left(\int x p(x) dx - \mu\right) - \lambda_2\left(\int (x-\mu)^2 p(x) dx - \sigma^2\right)$$Taking the functional derivative with respect to $p(x)$ and setting it to zero:$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - (\lambda_0 - 1) - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$$$p(x) = \exp\left(-\lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2\right)$$Because the distribution is centered on the mean constraint, the asymmetry parameter $\lambda_1$ must equal $0$. The function simplifies to:$$p(x) = e^{-\lambda_0} \exp\left(-\lambda_2 (x - \mu)^2\right)$$To satisfy the variance constraint $\sigma^2 = \int (x-\mu)^2 p(x) dx$, $\lambda_2$ must evaluate to $\frac{1}{2\sigma^2}$. Finally, to satisfy the normalization constraint $\int p(x) dx = 1$, the constant term $e^{-\lambda_0}$ must evaluate to $\frac{1}{\sqrt{2\pi\sigma^2}}$. Thus, the Maximum Entropy distribution is the Normal distribution:$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$3.
