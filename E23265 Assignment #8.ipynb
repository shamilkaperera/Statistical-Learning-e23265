### 1. Statistical Mechanics of the Quantum Oscillator

**1. The Partition Function**
The canonical partition function is given by:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{E_j}{k_B T}}$$

Substituting the energy eigenvalues $E_j = \hbar \omega \left(j + \frac{1}{2}\right)$ and defining $\beta = \frac{1}{k_B T}$:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar \omega \left(j + \frac{1}{2}\right)} = e^{-\frac{\beta \hbar \omega}{2}} \sum_{j=0}^{\infty} \left(e^{-\beta \hbar \omega}\right)^j$$

Since $e^{-\beta \hbar \omega} < 1$, the summation is a convergent infinite geometric series taking the form $\sum_{j=0}^{\infty} r^j = \frac{1}{1 - r}$.

$$Z(T) = e^{-\frac{\beta \hbar \omega}{2}} \left( \frac{1}{1 - e^{-\beta \hbar \omega}} \right) = \frac{e^{-\frac{\hbar \omega}{2k_B T}}}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

**2. Thermodynamic Potentials**

* **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$:

$$F(T) = -k_B T \left( -\frac{\hbar \omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right)$$

$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

* **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$:

$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$

$$U(T) = -\frac{\partial}{\partial \beta} \left( -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right) \right) = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}}$$

$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

**3. Entropy**
Using $S = -\left(\frac{\partial F}{\partial T}\right)$ and taking the derivative of $F(T)$:

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \left( \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \right) \left( -e^{-\frac{\hbar \omega}{k_B T}} \right) \left( \frac{\hbar \omega}{k_B T^2} \right)$$

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \left( \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right)$$

*Verification of $U = F + TS$:*

$$F + TS = \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)} \right]$$

$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U(T)$$

**4. Heat Capacity and Hyperbolic Form**

* **Heat Capacity**, $C_V$:

$$C_V = \frac{\partial U}{\partial T} = \frac{\partial}{\partial T} \left( \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right) = -\frac{\hbar \omega}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} \left( e^{\frac{\hbar \omega}{k_B T}} \right) \left( -\frac{\hbar \omega}{k_B T^2} \right)$$

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

* **Hyperbolic Form**: Multiply numerator and denominator by $e^{-\frac{\hbar \omega}{k_B T}}$:

$$\frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} = \frac{1}{e^{-\frac{\hbar \omega}{k_B T}}\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2} = \frac{1}{\left(e^{\frac{\hbar \omega}{2k_B T}} - e^{-\frac{\hbar \omega}{2k_B T}}\right)^2}$$

Since $\sinh(x) = \frac{e^x - e^{-x}}{2}$, we know $(e^x - e^{-x})^2 = 4 \sinh^2(x)$. Thus:

$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{4 \sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)} = k_B \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_B T}\right)}$$

**5. Asymptotic Behavior (Discussion)**
In the high-temperature limit ($k_B T \gg \hbar \omega$), we define $x = \frac{\hbar \omega}{2k_B T}$ where $x \to 0$. Using the Taylor expansion $\sinh(x) \approx x$ for small $x$:

$$C_V \approx k_B x^2 \frac{1}{x^2} = k_B$$

Yes, this exactly matches the classical equipartition theorem prediction. A classical 1D harmonic oscillator has two quadratic degrees of freedom (kinetic and potential energy), each contributing $\frac{1}{2}k_B$ to the heat capacity, yielding a total of $C_V = k_B$.

### 2. The Maximum Entropy Distribution under Mean and Variance Constraints

We want to find $p(x)$ that maximizes the differential entropy $H = -\int p(x) \ln p(x) dx$ subject to three constraints:

1. $\int p(x) dx = 1$
2. $\int x p(x) dx = \mu$
3. $\int (x-\mu)^2 p(x) dx = \sigma^2$

We set up the Lagrangian with multipliers $\lambda_0, \lambda_1, \lambda_2$:

$$\mathcal{L} = -\int p(x) \ln p(x) dx - (\lambda_0 - 1)\left(\int p(x) dx - 1\right) - \lambda_1\left(\int x p(x) dx - \mu\right) - \lambda_2\left(\int (x-\mu)^2 p(x) dx - \sigma^2\right)$$

Taking the functional derivative with respect to $p(x)$ and setting it to zero:

$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - (\lambda_0 - 1) - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

$$p(x) = \exp\left(-\lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2\right)$$

Because the distribution is centered on the mean constraint, the asymmetry parameter $\lambda_1$ must equal $0$. The function simplifies to:

$$p(x) = e^{-\lambda_0} \exp\left(-\lambda_2 (x - \mu)^2\right)$$

To satisfy the variance constraint $\sigma^2 = \int (x-\mu)^2 p(x) dx$, $\lambda_2$ must evaluate to $\frac{1}{2\sigma^2}$. Finally, to satisfy the normalization constraint $\int p(x) dx = 1$, the constant term $e^{-\lambda_0}$ must evaluate to $\frac{1}{\sqrt{2\pi\sigma^2}}$. Thus, the Maximum Entropy distribution is the Normal distribution:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

### 3. Multiclass Logistic Regression: Automatic Classification of Emails

The Maximum Entropy principle applied to discrete classification yields the Logistic Regression (Softmax) functional form. For a binary target output $Y \in \{0, 1\}$ (Finance vs. Technical) given feature vector $\mathbf{x} = (x_1, x_2, x_3)^T$, the probability distribution takes the form:

$$P(Y=1 \mid \mathbf{x}) = \frac{\exp(w_0 + w_1 x_1 + w_2 x_2 + w_3 x_3)}{1 + \exp(w_0 + w_1 x_1 + w_2 x_2 + w_3 x_3)}$$

To find the specific model parameters $\mathbf{w}$, the empirical expectations (Constraints 1-3) must strictly equal the model's expectations:

$$\mathbb{E}_{\text{model}}[x_i \cdot \mathbb{I}(Y=1)] = \sum_{\mathbf{x} \in \mathscr{X}} P(\mathbf{x}) P(Y=1 \mid \mathbf{x}) x_i$$

Based on the scenarios $\mathscr{X} \in \{\mathbf{x}_A, \mathbf{x}_B, \mathbf{x}_C, \mathbf{x}_D, \mathbf{x}_E\}$, this produces the following system of non-linear constraint equations:

**Constraint 1 ($x_1$):**

$$P(\mathbf{x}_A) P(Y=1 \mid \mathbf{x}_A)(1) + P(\mathbf{x}_E) P(Y=1 \mid \mathbf{x}_E)(1) = 0.05$$

$$0.40 \cdot \sigma(w_0 + w_1) + 0.05 \cdot \sigma(w_0 + w_1 + w_2) = 0.05$$

**Constraint 2 ($x_2$):**

$$P(\mathbf{x}_B) P(Y=1 \mid \mathbf{x}_B)(1) + P(\mathbf{x}_D) P(Y=1 \mid \mathbf{x}_D)(1) + P(\mathbf{x}_E) P(Y=1 \mid \mathbf{x}_E)(1) = 0.15$$

$$0.25 \cdot \sigma(w_0 + w_2) + 0.15 \cdot \sigma(w_0 + w_2 + w_3) + 0.05 \cdot \sigma(w_0 + w_1 + w_2) = 0.15$$

**Constraint 3 ($x_3$):**

$$P(\mathbf{x}_C) P(Y=1 \mid \mathbf{x}_C)(1) + P(\mathbf{x}_D) P(Y=1 \mid \mathbf{x}_D)(1) = 0.16$$

$$0.05 \cdot \sigma(w_0 + w_3) + 0.15 \cdot \sigma(w_0 + w_2 + w_3) = 0.16$$

*(Where $\sigma(z) = \frac{1}{1 + e^{-z}}$. Resolving this system using iterative numerical methods like Gradient Descent or Newton-Raphson yields the optimal weights).*

### 4. Asset Pricing and Portfolio Optimization

By the Principle of Maximum Entropy, we maximize $H = -\int_{0}^{\infty} p(S_T) \ln p(S_T) dS_T$ subject to the given market constraints:

1. $\int p(S_T) dS_T = 1$
2. $\int S_T p(S_T) dS_T = 102$
3. $\int S_T^2 p(S_T) dS_T = 10600$
4. $\int \ln(S_T) p(S_T) dS_T = 4.60$

Applying the calculus of variations via Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2, \lambda_3$, the probability density function takes the exponential family form:

$$p(S_T) = \exp\left( -\lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) \right)$$

Rewriting by factoring out $\lambda_3 \ln(S_T)$ as a base and absorbing $\lambda_0$ into a normalization constant $Z$:

$$p(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left( -\lambda_1 S_T - \lambda_2 S_T^2 \right)$$

This functionally describes a continuous generalized density distribution over $(0, \infty)$ where the parameters $\lambda_1, \lambda_2, \lambda_3$ and the partition function $Z$ are solved strictly to fit the macroeconomic aggregate constants ($102$, $10600$, and $4.60$). Because it explicitly incorporates the $\ln(S_T)$ constraint, it accounts for the heavy downside tail (skew) without defaulting to the rigid Lognormal or Gaussian assumptions of standard Black-Scholes.
