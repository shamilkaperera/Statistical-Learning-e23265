Q9) Information Available in a Discrete Random Variable
$$P(\{G_1\})=0.50$, $P(\{G_2\})=0.20$, $P(\{D_1\})=0.10$, $P(\{D_2\})=0.20.$$
1. Probability space:
The sample space is $\Omega = \{G_1, G_2, D_1, D_2\}$. The $\sigma$-algebra $\mathfrak{F} = \mathfrak{P}(\Omega)$ is the power set

2. $\sigma$-algebra of $X$:
$X$ groups the outcomes by quality (0 for good, 1 for defective). The pre-images are $X^{-1}(0) = \{G_1, G_2\}$ and $X^{-1}(1) = \{D_1, D_2\}$.

3. $\sigma$-algebra of $Y$:
$Y$ assigns a unique value to every single outcome. Thus, every singleton set is measurable.
  $$\sigma(Y) = \mathfrak{P}(\Omega)$$

  4. Because every set in $\sigma(X)$ is also present in $\sigma(Y)$, it is true that $\sigma(X) \subseteq \sigma(Y)$. $Y$ contains more information because knowing the value of $Y$ leaves zero uncertainty about the precise outcome, whereas knowing $X$ only tells you the quality, leaving you uncertain about the machine used

  5. Marginal distribution of $X$:$P(X=0) = P(\{G_1, G_2\}) = 0.50 + 0.20 = 0.70$$P(X=1) = P(\{D_1, D_2\}) = 0.10 + 0.20 = 0.30$

  6. Marginal distribution of $Y$:
$P(Y=1) = 0.50$, $P(Y=2) = 0.20$, $P(Y=3) = 0.10$, $P(Y=4) = 0.20$.

7. Joint probabilities: $$P(X=0, Y=1) = P(\{G_1\}) = 0.50$$$$P(X=0, Y=2) = P(\{G_2\}) = 0.20$$$$P(X=1, Y=3) = P(\{D_1\}) = 0.10$$$$P(X=1, Y=4) = P(\{D_2\}) = 0.20$$


8. Conditional probabilities:$$P(X=1 \mid Y=3) = \frac{P(X=1, Y=3)}{P(Y=3)} = \frac{0.10}{0.10} = 1$$$$P(X=1 \mid Y=4) = \frac{P(X=1, Y=4)}{P(Y=4)} = \frac{0.20}{0.20} = 1$$$$P(Y=3 \mid X=1) = \frac{P(X=1, Y=3)}{P(X=1)} = \frac{0.10}{0.30} = \frac{1}{3}$$$$P(Y=4 \mid X=1) = \frac{P(X=1, Y=4)}{P(X=1)} = \frac{0.20}{0.30} = \frac{2}{3}$$

9. Interpret conditioning difference:
Conditioning on $Y$ means we fully know the outcome. Consequently, $X$ is perfectly deterministic. Conditioning on $X$ means we only know the quality.

10. Explaining Information in Words:
$\sigma(X)$ contains information about the product's overall quality . $\sigma(Y)$ contains complete diagnostic information: both the quality and the exact machine source. All information in $X$ is present in $Y$, but the machine origin information is unique to $\sigma(Y)$.



```python
import matplotlib.pyplot import plt
import seaborn as sns
import numpy as np

# Joint probability matrix P(X, Y)
# Rows: X=0, X=1 | Cols: Y=1, Y=2, Y=3, Y=4
joint_probs = np.array([
    [0.50, 0.20, 0.0, 0.0],
    [0.0, 0.0, 0.10, 0.20]
])

sns.heatmap(joint_probs, annot=True, cmap="Blues",
            xticklabels=['Y=1', 'Y=2', 'Y=3', 'Y=4'],
            yticklabels=['X=0', 'X=1'])
plt.title("Joint Probability Heatmap P(X,Y)")
plt.xlabel("Y (Machine & Quality)")
plt.ylabel("X (Quality)")
plt.show()
```

Q10) Information Available in Continuous Random Variables

$T \sim N(0,1)$ and $P \sim N(0,4)$ measuring reactor deviations.

1. Probability Space
$P$ is the measure defined by integrating the joint density:
$$f_{T,P}(t,p) = \frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right)$$

2. $\sigma$-algebra of X
Because $X(t,p) = t$, its $\sigma$-algebra consists of "vertical strips." It is $\sigma(X) = \{A \times \mathbb{R} \mid A \in \mathfrak{B}(\mathbb{R})\}$.

3. $\sigma$-algebra of Y
Because $Y(t,p) = (t,p)$ is the identity map, it generates the full Borel $\sigma$-algebra of the space. $\sigma(Y) = \mathfrak{B}(\mathbb{R}^2)$.

4. Show $\sigma(X) \subseteq \sigma(Y)$
Every set of the form $A \times \mathbb{R}$ (from $\sigma(X)$) is inherently a Borel set in $\mathbb{R}^2$ (from $\sigma(Y)$). $Y$ has more information because it distinguishes points varying in the $p$ coordinate, whereas $X$ merges all points with the same $t$ coordinate.

5. Marginal Density of X
Since $X$ is just the temperature $T \sim N(0,1)$:
$$f_X(x) = \frac{1}{\sqrt{2\pi}} \exp\left(-\frac{x^2}{2}\right)$$

6. Marginal Density of Y
Since $Y = (T,P)$, its density is the joint density:
$$f_Y(t,p) = \frac{1}{4\pi}\exp\left(-\frac{t^2}{2}-\frac{p^2}{8}\right)$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Meshgrid for joint density
t = np.linspace(-3, 3, 100)
p = np.linspace(-6, 6, 100)
T, P = np.meshgrid(t, p)
Z = (1/(4*np.pi)) * np.exp(-T**2/2 - P**2/8)

plt.figure(figsize=(10, 4))

# 1. Joint density contour Y=(T,P)
plt.subplot(1, 2, 1)
plt.contourf(T, P, Z, levels=20, cmap='viridis')
plt.title("Joint Density of Y = (T, P)")
plt.xlabel("Temperature (T)")
plt.ylabel("Pressure (P)")

# 2. Marginal density of X (T)
plt.subplot(1, 2, 2)
X_marg = (1/np.sqrt(2*np.pi)) * np.exp(-t**2/2)
plt.plot(t, X_marg, color='red')
plt.title("Marginal Density of X (T)")
plt.xlabel("Temperature (T)")

plt.tight_layout()
plt.show()
```

8. Conditional distribution of X given Y=(t,p)
Because $Y$ fully specifies $X$, the distribution collapses to a Dirac delta distribution centered precisely at $t$, $\delta(x-t)$.

9. Conditional distribution of Y given X=t
Since $X$ fixes the $T$ component to $t$ and $P$ is independent of $T$, the remaining uncertainty lies entirely in $P$. The distribution is $\{t\} \times N(0,4)$.

10. Explaining in Words
Conditioning on $X$ means we observe the exact temperature deviation but remain completely ignorant about the pressure (it maintains its prior $N(0,4)$ distribution). Conditioning on $Y$ reveals the exact sensor state for both variables, resolving all uncertainty.

11. Show $X = \pi_1 \circ Y$
The projection mapping is $\pi_1(t,p) = t$.\\
Applying this to $Y$, we get $\pi_1(Y(t,p)) = \pi_1(t,p) = t = X(t,p)$. Because $X$ is a composition of a measurable function ($\pi_1$) with $Y$, it guarantees $X$ is $\sigma(Y)$-measurable, mandating $\sigma(X) \subseteq \sigma(Y)$.


13. Differential entropy of X
For a normal distribution with variance $\sigma^2$, $h = \frac{1}{2}\ln(2\pi e \sigma^2)$. Since $X \sim N(0,1)$:
$$h(X) = \frac{1}{2}\ln(2\pi e)$$

14. Differential entropy of Y
For a bivariate normal with independent components, the joint entropy is the sum of the marginal entropies. We know $h(P) = \frac{1}{2}\ln(2\pi e \times 4) = \frac{1}{2}\ln(8\pi e)$.
$$h(Y) = h(T) + h(P) = \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e) = \frac{1}{2}\ln(16\pi^2 e^2) = \ln(4\pi e)$$

15. Conditional differential entropy $h(Y \mid X)$
$$h(Y \mid X) = h(Y) - h(X) = \ln(4\pi e) - \frac{1}{2}\ln(2\pi e) = \frac{1}{2}\ln(8\pi e)$$
Physical interpretation: Since temperature and pressure are independent, observing the temperature ($X$) removes no uncertainty regarding the pressure. The remaining entropy is exactly $h(P)$.

16. Discuss $h(X \mid Y)$
Because $Y$ fully determines $X$, the remaining uncertainty is zero. In the discrete case, $H = 0$. However, in the continuous case, the differential entropy of a deterministic or perfectly known variable diverges to $-\infty$
